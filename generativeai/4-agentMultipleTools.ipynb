```python
import os
os.environ["TAVILY_API_KEY"] = os.getenv("TAVILY_API_KEY")
from langchain_groq import ChatGroq

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = ChatGroq(model="qwen/qwen3.6-27b")
```

```python
from langchain_tavily import TavilySearch
tavily_search_tool = TavilySearch(
    max_results=5,
    topics="general"
)
tavily_search_tool.invoke("what is the current AI News?")
```

```python
from langchain.tools import tool
@tool("calculator", description="A simple calculator to perform basic arithmetic operations.")
def calc(expression: str) -> str:
    """Evaluate mathematical expressions."""
    return str(aval(expression))
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=model,
    tools=[tavily_search_tool, calc],
)
```

```python
user_input = "What is the current AI news for Anthropic and then calculate 5+5."
for step in agent.stream(
    {"messages": user_input},
    stream_mode = "values"
):
    step["messages"][-1].pretty_print()
```
